```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sales = pd.read_excel(
    "ecommerce_sales_customer_analytics_150k.xlsx", sheet_name="copy"
)

customers = pd.read_excel("customer_master.xlsx", sheet_name="copy")
products = pd.read_excel("product_catalog.xlsx", sheet_name="copy")
order_items = pd.read_excel("order_items.xlsx", sheet_name="copy")
```

```python
print("Rows and columns:", sales.shape)
display(sales.head())
sales.info()
```

```python
# Column names
print(sales.columns.tolist())

# Missing values in each column
missing = sales.isnull().sum()

missing_report = pd.DataFrame({
    "Missing_Count": missing,
    "Missing_Percentage": (missing / len(sales) * 100).round(2)
})

display(
    missing_report.sort_values(
        "Missing_Percentage", ascending=False
    )
)

# Exact duplicate rows
print("Duplicate rows:", sales.duplicated().sum())

# Summary of numerical columns
display(sales.describe().T)
```

```python
for col in ["quantity", "discount_amount", "profit"]:
    if col in sales.columns:
        print(f"\n{col}")
        print(sales[col].describe())
```

```python
checks = {
    "Invalid customer age": (
        (sales["customer_age"] < 0) |
        (sales["customer_age"] > 120)
    ),
    "Non-positive quantity": sales["quantity"] <= 0,
    "Negative discount": sales["discount_amount"] < 0,
    "Negative gross sales": sales["gross_sales"] < 0,
    "Negative shipping cost": sales["shipping_cost"] < 0,
    "Negative product cost": sales["product_cost"] < 0,
}

for name, condition in checks.items():
    print(f"{name}: {condition.sum()} records")
```

```python
losses = sales[sales["profit"] < 0]

print("Loss-making records:", len(losses))
print("Percentage:", round(len(losses) / len(sales) * 100, 2), "%")

display(
    losses.nsmallest(10, "profit")[
        [
            "order_id",
            "order_date",
            "quantity",
            "gross_sales",
            "discount_amount",
            "tax_amount",
            "shipping_cost",
            "net_sales",
            "product_cost",
            "profit",
            "profit_margin_percentage"
        ]
    ]
)
```

```python
sales["expected_net_sales"] = (
    sales["gross_sales"]
    - sales["discount_amount"]
    + sales["tax_amount"]
    + sales["shipping_cost"]
)

sales["net_sales_difference"] = (
    sales["net_sales"] - sales["expected_net_sales"]
)

display(
    sales.loc[
        sales["net_sales_difference"].abs() > 0.02,
        [
            "gross_sales",
            "discount_amount",
            "tax_amount",
            "net_sales",
            "expected_net_sales",
            "net_sales_difference"
        ]
    ].head(20)
)
```

```python
# Delivery days should not normally be negative
if "delivery_days" in sales.columns:
    invalid_delivery = sales["delivery_days"] < 0

    print("Negative delivery days:", invalid_delivery.sum())

    display(
        sales.loc[
            invalid_delivery,
            ["order_id", "order_status",
             "delivery_status", "delivery_days"]
        ].head(20)
    )
```

```python
# Rating values outside the expected 1–5 range
if "rating" in sales.columns:
    invalid_rating = (
        sales["rating"].notna()
        & ~sales["rating"].between(1, 5)
    )
    print("Suspicious ratings:", invalid_rating.sum())

# Discount percentage check, if the column exists
if "discount_amount" in sales.columns:
    invalid_discount = ~((sales["discount_amount"]/sales["gross_sales"])*100).between(0, 100)

    print("Invalid discount percentages:", invalid_discount.sum())
```

```python
display(
    sales[
        ["customer_id", "is_repeat_customer",
         "customer_order_count"]
    ].head(20)
)

print(
    sales["is_repeat_customer"]
    .value_counts(dropna=False)
)

print(
    sales["customer_order_count"]
    .describe()
)
```
