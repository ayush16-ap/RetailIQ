```python
import pandas as pd
import numpy as np

customers = pd.read_csv("E:\CODING_ALL\Data Analytics\projrcts\RetailIQ\data\Cleaned csv\cleaned_customers.csv")
features = pd.read_csv("customer_features.csv")
rfm = pd.read_csv("customer_rfm_segments.csv")
kmeans = pd.read_csv("customer_kmeans_segments.csv")

print("Customers:", customers.shape)
print("Features:", features.shape)
print("RFM:", rfm.shape)
print("K-Means:", kmeans.shape)
```

```python
customer_info = customers[
    ["customer_id", "customer_name", "customer_segment",
     "customer_city", "customer_state", "region"]
].copy()

feature_cols = [
    "customer_id",
    "recency_days",
    "frequency",
    "monetary_value",
    "avg_order_value",
    "avg_discount_rate",
    "return_rate",
    "avg_delivery_days",
    "purchase_frequency"
]

customer_features = features[
    [col for col in feature_cols if col in features.columns]
].copy()

rfm_cols = [
    "customer_id", "R_score", "F_score", "M_score",
    "RFM_score", "customer_group"
]

rfm_results = rfm[
    [col for col in rfm_cols if col in rfm.columns]
].copy()

kmeans_cols = [
    "customer_id", "cluster", "cluster_label"
]

kmeans_results = kmeans[
    [col for col in kmeans_cols if col in kmeans.columns]
].copy()
```

```python
customer_segments = (
    customer_info
    .merge(customer_features, on="customer_id", how="left",
           suffixes=("", "_feature"))
    .merge(rfm_results, on="customer_id", how="left")
    .merge(kmeans_results, on="customer_id", how="left")
)

print("Final shape:", customer_segments.shape)
customer_segments.head()
```

```python
print("Unique customer IDs:", customer_segments["customer_id"].nunique())
print("Total rows:", len(customer_segments))
print("Duplicate customer IDs:",
      customer_segments["customer_id"].duplicated().sum())

print("\nMissing values:")
print(customer_segments.isnull().sum())

assert customer_segments["customer_id"].is_unique, (
    "Duplicate customer IDs found. Investigate before exporting."
)

assert customer_segments["customer_id"].notna().all(), (
    "Missing customer IDs found."
)
```

```python
customer_segments.to_csv(
    "customer_segments.csv",
    index=False
)

print("Export completed: customer_segments.csv")
print("Rows:", len(customer_segments))
print("Columns:", len(customer_segments.columns))
```
