```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
```

```python
product = pd.read_excel(
    r"E:\CODING_ALL\Data Analytics\projrcts\RetailIQ\data\processed\product_catalog.xlsx",
    sheet_name="processed"
)

df = product.copy()
```

```python
print("Dataset shape:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

display(df.head())
```

```python
df.columns = df.columns.str.strip()

print("Dataset shape:", df.shape)
print("\nData types:")
print(df.dtypes)

numeric_cols = [
    "unit_price",
    "product_cost",
    "product_rating"
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

display(df[numeric_cols].describe().T)
```

```python
missing = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (df.isnull().mean() * 100).round(2)
})

display(missing.sort_values("Missing_Percentage", ascending=False))

print("Exact duplicate rows:", df.duplicated().sum())
print("Missing product IDs:", df["product_id"].isna().sum())
print("Unique product IDs:", df["product_id"].nunique())
print("Duplicate product IDs:", df["product_id"].duplicated().sum())
```

```python
checks = {
    "Negative unit price": df["unit_price"] < 0,
    "Zero unit price": df["unit_price"] == 0,
    "Negative product cost": df["product_cost"] < 0,
    "Negative product rating": df["product_rating"] < 0,
    "Rating above 5": df["product_rating"] > 5
}

for issue, condition in checks.items():
    print(f"{issue}: {condition.sum()}")
```

```python
categorical_cols = [
    "product_name",
    "product_category",
    "product_subcategory",
    "brand",
    "supplier"
]

for col in categorical_cols:
    print(f"\n--- {col} ---")
    print("Unique values:", df[col].nunique())
    print("Missing values:", df[col].isna().sum())
    print(df[col].value_counts(dropna=False).head(10))
```

```python
df["price_cost_difference"] = df["unit_price"] - df["product_cost"]

print("Products priced below cost:",
      (df["price_cost_difference"] < 0).sum())

display(
    df.loc[
        df["price_cost_difference"] < 0,
        ["product_id", "product_name", "unit_price", "product_cost"]
    ].head(20)
)
```

```python
df.to_csv("cleaned_product.csv", index=False)
print("Done!")
```
