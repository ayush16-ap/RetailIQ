# Order Items Data Exploration and Cleaning

## 1. Dataset Overview

```python
# Dataset overview
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())

display(df.head())

print("\nData types:")
df.info()
```

## 2. Statistical Analysis


```python
# Statistical summary
display(df.describe(include="all").T)
```

## 3. Data Quality Analysis


```python
# Missing values
missing_report = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": (
        df.isnull().mean() * 100
    ).round(2)
})

display(
    missing_report.sort_values(
        "Missing_Percentage",
        ascending=False
    )
)
```

```python
# Duplicate analysis
print("Exact duplicate rows:", df.duplicated().sum())
```

## 4. Data Cleaning

```python
# Clean column names
df.columns = df.columns.str.strip()

# Clean text columns
text_cols = df.select_dtypes(
    include=["object", "string"]
).columns

for col in text_cols:
    df[col] = df[col].str.strip()
    df[col] = df[col].replace(
        r"^\s*$", np.nan, regex=True
    )

print("Basic text cleaning completed")
```

```python
print(df.columns.tolist())
print(df.select_dtypes(include="number").describe().T)
```

```python
numeric_cols = [
    "quantity",
    "unit_price",
    "discount_percentage",
    "discount_amount",
    "gross_sales",
    "tax_amount",
    "shipping_cost",
    "net_sales",
    "product_cost",
    "profit"
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print(df[numeric_cols].describe().T)
```

## 5. Validation

```python
print("Missing values:")
display(df.isnull().sum().to_frame("Missing_Count"))

print("Exact duplicate rows:", df.duplicated().sum())

print("Missing order IDs:", df["order_id"].isna().sum())
print("Missing product IDs:", df["product_id"].isna().sum())
```

```python
checks = {
    "Quantity <= 0": df["quantity"] <= 0,
    "Negative unit price": df["unit_price"] < 0,
    "Discount percentage outside 0-100": (
        ~df["discount_percentage"].between(0, 100)
    ),
    "Negative discount amount": df["discount_amount"] < 0,
    "Negative gross sales": df["gross_sales"] < 0,
    "Negative tax": df["tax_amount"] < 0,
    "Negative shipping cost": df["shipping_cost"] < 0,
    "Negative product cost": df["product_cost"] < 0
}

for issue, condition in checks.items():
    print(f"{issue}: {condition.sum()}")
```

```python
df["expected_gross_sales"] = (
    df["quantity"] * df["unit_price"]
)

df["gross_sales_difference"] = (
    df["gross_sales"] - df["expected_gross_sales"]
)

display(
    df.loc[
        df["gross_sales_difference"].abs() > 0.02,
        [
            "order_id",
            "product_id",
            "quantity",
            "unit_price",
            "gross_sales",
            "expected_gross_sales",
            "gross_sales_difference"
        ]
    ].head(20)
)
```
